```python
class BankAccount:
    
    def __init__(self):
        self.name = input("please enter your name: ")
        self.__nationalid = input("please enter id: ")
        self.__password = int(input("please enter your password"))
        self.__phonenumber = input("please enter phone: ")
        self.__credit = 0 
    
    def deposit(self):
        amount= int(input('please enter your deposit amount: '))
        self.__credit += amount
        print("Done, balance has been added.")
    
    def withdraw(self): 
        accountpass = int(input("please enter your password: "))
        if accountpass == self.__password: 
            amount = int(input('please enter your withdraw amount: '))
            if amount <= self.__credit: 
                self.__credit -= amount
                print("Done, please take your money")
            else:
                print("sorry, you're poor ")
        else:
            print("wrong password")
    
    def checkcredit(self): # getter ( access private data ) getcredit , get_credit
        accountpass = int(input("please enter your password: "))
        if accountpass == self.__password: 
            print(self.__credit)
        else:
            print("Wrong password")
    
    def changepassword(self): # setter ( access private data) setcredit , set_credit
        accountpass = int(input("please enter your password: "))
        if accountpass == self.__password: 
            newpassword = int(input("please enter your new password: "))
            self.__password = newpassword
            print("password changed successfully")
        else:
            print("Wrong password")
```

### Composition 

```python
class Composition: 
    
    def __init__(self):
        self.__att = dict()
        self.__methods = dict()

    def addatt(self, attname, attvalue):
        self.__att[attname] = attvalue
    
    def addmethod(self, methodname , method):
        self.__methods[methodname] = method
        
    
    def editatt(self, attname , newvalue):
        self.__att[attname] = newvalue
    
    def deleteatt(self, attname):
        del self.__att[attname]
    
    def deletemethod(self, methodname):
        del self.__methods[methodname]
        
    def printatt(self, attname):
        print(self.__att[attname])
    
    def printallatt(self):
        for attribute in self.__att:
            print(attribute)
    
    def printallmethods(self):
        for m in self.__methods:
            print(m)
    
    def callmethod(self, methodname):
        self.__methods[methodname]()
```
