```python
class Composition:


    def __init__(self):
        self.__attributes = dict()
        self.__methods = dict()


    def att_attribute(self, att_name , att_value): 
        self.__attributes[att_name] = att_value

    def edit_attribute(self, att_name , new_value): 
        self.__attributes[att_name] = new_value

    def delete_attribute(self, att_name): 
        del self.__attributes[att_name]

    def print_attribute(self, att_name): 
        print( self.__attributes[att_name])

    def print_all_attributes(self): 
        if len(self.__attributes) == 0: 
            print("there are no attributes")
        for attribute in self.__attributes: 
            print(attribute)

    def add_method(self, method_name, method):
        self.__methods[method_name] = method

    def delete_method(self, method_name):
        del self.__methods[method_name]

    def call_method(self, method_name):
        self.__methods[method_name]()

    def print_all_methods(self):
        print( "Available methods to use: ")
        for method in self.__methods:
            print(method)
```
