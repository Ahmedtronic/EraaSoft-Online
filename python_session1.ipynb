```python
def print_info( *data ): 
    print(data[0])
    print(data[1])
```

```python
print_info("Ahmed", "Ali", "Essam")
```

```python
print_info("Ahmed")
```

```python
print_info("Ahmed", "Ali", "Essam")
```

```python
def print_info( *data ): 
    if "Ahmed" in data: 
        print("yes")
```

```python
print_info("essam")
```

```python
def print_info( **data): 
    print(data["name"])
    print(data["age"])
```

```python
print_info(name = "Ahmed", age = 23 )
```

```python
print_info(age = 23, name = "Ahmed"  )
```

```python
print_info(age = 23, name = "Ahmed" , salary = 1000  )
```

```python
info = { 
    "name" : "Ahmed", 
    "age" : 23 , 
    "salary" : 1000 
}
```

```python
print_info(**info)
```
