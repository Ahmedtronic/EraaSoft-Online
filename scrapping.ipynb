```python
for i in range(1,11): 
    url = f"https://quotes.toscrape.com/page/{i}/" 
    website = requests.get(url).text
    html_website = BeautifulSoup(website , "html")
    quotes = html_website.find_all("div", class_= "quote")
    for quote in quotes:
        quote_text = quote.find("span" , class_= "text").text 
        quote_person = quote.find("small" , class_= "author").text 
        print(quote_text , "by:" , quote_person )
```

```python
today_matches = "https://www.yallakora.com/matches-center#matchesclipAll"
website = requests.get(today_matches).text
```

```python
print(website)
```

```python
html_website = BeautifulSoup(website, "html")
```

```python
first_match = html_website.find("div", class_ = "allData")
first_match
```

```python
o1 = first_match.find("div", class_ = "teams teamA").p.text
o1
```

```python
o2 = first_match.find("div", class_ = "teams teamB").p.text
o2
```

```python
results = first_match.find("div", class_= "MResult" )
results_number = results.find_all("span", class_ = "score")
result_values = [ ]
for result in results_number : 
    result_values.append(result.text)

result_o1 = result_values[0]
result_o2 = result_values[1]

result_o1 , result_o2
```

```python
matches = html_website.find_all("div", class_ = "allData")
for match in matches: 
    o1 = match.find("div", class_ = "teams teamA").p.text
    o2 = match.find("div", class_ = "teams teamB").p.text
    results = match.find("div", class_= "MResult" )
    results_number = results.find_all("span", class_ = "score")
    result_values = [ ]
    for result in results_number : 
        result_values.append(result.text)

    result_o1 = result_values[0]
    result_o2 = result_values[1]

    print("Team A:", o1 , "Goals:" , result_o1 )
    print("Team B:", o2 , "Goals:" , result_o2 )
    print("==========================")
    


```
