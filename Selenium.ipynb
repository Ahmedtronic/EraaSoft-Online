```python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

driver = webdriver.Firefox()
driver.get("https://www.google.com")
time.sleep(2)

# Find the search box and type a query
search_box = driver.find_element(By.NAME, "q") # find element = find tag which has attribute called name ans its value is q 
search_box.send_keys("Selenium Python")
time.sleep(3)
search_box.send_keys(Keys.RETURN) # Pressed Enter 

# Wait and get results
time.sleep(5)
results = driver.find_elements(By.CSS_SELECTOR, "h3")
for r in results[:5]:
    print(r.text)
```

```python
driver = webdriver.Firefox()
driver.get("https://www.w3schools.com/html/html_forms.asp")

# Try filling a form (example form)
first_name = driver.find_element(By.ID, "fname")
last_name = driver.find_element(By.ID, "lname")
time.sleep(1)
first_name.clear()
time.sleep(2)
first_name.send_keys("Ahmed")
time.sleep(2)
last_name.clear()
time.sleep(2)
last_name.send_keys("Hafez")
time.sleep(3)

# Click the submit button
submit_btn = driver.find_element(By.CSS_SELECTOR, "button[type='submit']")
time.sleep(5)
submit_btn.click()
time.sleep(2)

driver.quit()
```

### Using Selenium for Scrapping 

```python
driver = webdriver.Firefox()


driver.get("https://quotes.toscrape.com")
time.sleep(2)

try:
    # Get the first quote by class name
    quote = driver.find_element(By.CLASS_NAME, "text") # element = tag 
    print(quote)
    print("First quote at this website:", quote.text)
except Exception as e:
    print("Error:", e)

driver.quit()
```

```python
driver = webdriver.Firefox()


driver.get("https://quotes.toscrape.com")
time.sleep(5)


try:
    quotes = driver.find_elements(By.CLASS_NAME, "text")

    # Loop through and print each quote
    for i, quote in enumerate(quotes, 1):
        print(f"Quote {i} is: {quote.text}")

except Exception as e:
    print("Error:", e)

driver.quit()
```

<h2> Scrolling Through Web Page </h2> 

```python
driver = webdriver.Firefox()


driver.get("https://www.udemy.com/")
time.sleep(5)


# Run js Comman: Scroll down by 1000 pixels 
driver.execute_script("window.scrollBy(0, 1000);")


time.sleep(2)


driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

# Optional: Scroll to a specific element
# element = driver.find_element(By.ID, "element_id")
# driver.execute_script("arguments[0].scrollIntoView();", element)

# Close the driver
driver.quit()
```

```python
driver = webdriver.Firefox()

driver.get("https://www.udemy.com/")
time.sleep(5)


for i in range(25): 
    driver.execute_script("window.scrollBy(0, 100);")
    time.sleep(1)
    

driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(2) 
# Optional: Scroll to a specific element
# element = driver.find_element(By.ID, "element_id")
# driver.execute_script("arguments[0].scrollIntoView();", element)

driver.quit()
```

```python
driver = webdriver.Firefox()
driver.get("https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population")
time.sleep(10)

table = driver.find_element(By.CLASS_NAME, "wikitable")
rows = table.find_elements(By.TAG_NAME, "tr")

for row in rows[:17]: 
    cols = [c.text for c in row.find_elements(By.TAG_NAME, "td")]
    print(cols)

driver.quit()
```
